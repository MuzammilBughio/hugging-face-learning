```python
from diffusers import StableDiffusionPipeline
```

```python
import matplotlib.pyplot as plt
import torch
```

```python
!pip show torch
```

```python
from diffusers.utils.import_utils import USE_SAFETENSORS
model_id1 = "dreamlike-art/dreamlike-diffusion-1.0"
model_id2 = "stabilityai/stable-diffusion-xl-base-1.0"

pipe = StableDiffusionPipeline.from_pretrained(model_id1 ,torch_dtype=torch.float16 , use_safetensors= True)
pipe = pipe.to('cuda')
```

```python
prompt = """
          A lone traveler standing on a cliff above a futuristic cyberpunk city at sunset, glowing neon lights reflecting in the rain-soaked streets below, flying vehicles moving between towering skyscrapers, cinematic lighting, ultra-detailed environment, dramatic clouds, volumetric fog, realistic reflections, high contrast, photorealistic, 8k quality, masterpiece composition, depth of field, award-winning photography style.
          """
```

```python
image = pipe(prompt).images[0]
```

```python
plt.imshow(image);
plt.axis('off');
```
