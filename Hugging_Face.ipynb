```python
from transformers import pipeline
```

```python
classifier = pipeline('sentiment-analysis')
result = classifier("i am not so happy with Hero Movie")
print(result)
```

```python
classifier = pipeline(task = 'sentiment-analysis', model="SamLowe/roberta-base-go_emotions")
task_list = ['its so bad', \
             'its awfull', \
             'its lonely kitty ', \
             'its lovely kitty ', \
             'its nice']
classifier(task_list)
```

```python
classifier = pipeline(task= 'question-answering')
test = "What is LLM?"
classifier(test, max_length= 30, num_return_sequences=3)
```
